## Why and when we need a for loop?

When you want to do the same action on a sequence of things (items in a list, characters in a string, rows in a table, files in a folder), a for loop is Python’s way to say:

*“I will repeat this block once per item, without me rewriting the code N times.”*

You need a for loop when:

- You want to apply the same rule to each element.
Example: print each name, compute each square, check each value.
- You want to accumulate a result from many items.
Example: sum numbers, count how many meet a condition, build a new list based on a condition.

```python
 items = [
    0.23,
    5,
    "Berkeley",
    True,
    [1, 2, 3],
    (10, 20),
]

# What are the types of objects I have in my list?
# Without for loop

print(items[0], type(items[0]))
print(items[1], type(items[1]))
print(items[2], type(items[2]))
print(items[3], type(items[3]))
print(items[4], type(items[4]))
print(items[5], type(items[5]))
```

```python
print('*' * 20)

#with for loop
for item in items:
    print(item, type(item))
```

```python
# Defining groups
group_1 = ['Stefan', 'Maryam', 'Zoe', 'Daniel']


# Typing out all the names with ' and , can be annoying; str.split() helps.
group_2 = 'Claudio Joao Peter Neel Matt'.split()
group_3 = 'Deb Ewa Dustin'.split()

# Instead of handling each group separately, we put all groups into a list.

groups = [group_1, group_2, group_3]

# Sorting groups based on size in descending order.
# This ensures we handle larger groups first so that we have members to add to smaller groups if needed.
# you can check what other keys you can choose with sort!
groups.sort(key=len, reverse=True)

# printing len of groups without a for loop:
print(f"Group size is {len(groups[0])}")
print(f"Group size is {len(groups[1])}")
print(f"Group size is {len(groups[2])}")
```

### Accessing groups using a for loop and printing their length

When using for loop, you often do not need to use index to access each item
The for loop has direct access to each item in the iterable. It is like you have this privilidged access to look at any items just using python's `for` loop!

```python

for group in groups:
    print(f"Group size is {len(group)}")
```

```python
# If you need the index, use enumerate():
for i, group in enumerate(groups):
    print(f"Group {i} size is {len(group)}")
```

## Balancing Group Sizes

Let's say we have a strict group size rule. It needs to be four.
So we want to take each group and if the size doesnt match our rule, add or remove members to it.

```python
removed_members = []  # Stores extra members removed from oversized groups.
smaller_groups = []  # Tracks groups that still need more members.

for group in groups:
    if len(group) > 4:
        print(f"Group has {len(group)} members, exceeding 4.")
        removed_members.append(group.pop())  # Remove last member and store it
        print(f"Removed: {removed_members[-1]}")
        print(f"Updated group: {group}")

    elif len(group) < 4:
        print(f"Group has {len(group)} members, below 4.")
        if removed_members: # any nonzero value evaluate to True in Python, so if the list is not empty, this will pass.
            print(f"Adding {removed_members[-1]} to this group.")
            group.append(removed_members.pop())  # Add last removed member
            print(f"Updated group: {group}")
        else:
            print("No available members to add.")
            smaller_groups.append(group)  # Track underfilled groups

    else:
        print("Perfect group size.")
```
