##**What is debugging?**

Programming is a complex process. Code errors are called **bugs** and the process of finding and correcting them is called **debugging**.

Three kinds of errors can occur in a program:

- **Syntax errors:** Before running your code, Python will read it first and parse it to check the syntax. Python cannot read the code if it violates the rules of the language. A missing colon, an unclosed parenthesis, a misspelled keyword, can cuase Python to stop and prints an error message. It will refuse to run the code until these are fixed.

- **Runtime errors (exceptions):** the code is syntactically valid, but something goes wrong while the program is actually running. Dividing by zero, accessing a list index that does not exist, opening a file that is not there. Python stops at the point of failure and prints an error message telling you what happened and where.

- **Semantic errors:** the code runs from start to finish without any error messages, but it does not do what you intended. It does exactly what you *told* it to do -- which turns out to be different from what you *wanted* it to do. These are the hardest to find because there is no error message pointing you anywhere.

Some cells in this notebook are **intentionally written to produce errors**. That is normal!! The goal is for you to read and understand the error output.

##**Part 1: Errors That Crash Your Program**

### 1.1 How to Read an Error Message (Traceback)

When Python crashes, it prints a **traceback**: a message telling you what went wrong and where. They are structured and easy to read once you know the pattern. Run the cell below. It will produce an error.

````python
# This cell is supposed to error -- run it and look at the output
def compute_average(values):
    total = sum(values)
    count = len(values)
    return total / count

empty_list = []
result = compute_average(empty_list)
````

````python
nums = 2, 3, 5
compute_average(nums)
````

The output looks something like this:

```
Traceback (most recent call last):
  File "/path/to/notebook", line 8, in <module>
    result = compute_average(empty_list)
  File "/path/to/notebook", line 4, in compute_average
    return total / count
ZeroDivisionError: division by zero
```

**- Step 1: Start at the bottom.** The last line is always the most important:
```
ZeroDivisionError: division by zero
```
This tells you two things: the **error type** (`ZeroDivisionError`) and a **short description** of what went wrong (`division by zero`).


**- Step 2: Look at the line just above the error type.** This is the exact line of code that caused the crash:
```
    return total / count
```

**- Step 3: Look at the file and line number above that.** This tells you where in your code the problem is:
```
  File "/path/to/notebook", line 4, in compute_average
```

**- Step 4: Read upward if needed.** If the error happened inside a function, the traceback shows you the full call chain -- how the program got to that point. Reading upward shows you which function called which.

### 1.2 Common Python Error Types

Python has many error types, each corresponding to a different problem. The following cells in this section cover Syntax and Runtime errors. **Each cell is intentionally written to produce an error.** After you run it, read the traceback and read the explanation below.

---

####**1.2a - Syntax Errors:**

This happens before the program runs and *may not* show a full traceback.

<br>


#### **`SyntaxError:`** Python cannot read the code

A `SyntaxError` means Python could not even parse your code -- it found something that is not valid Python. This is the equivalent of a grammatical error in a sentence. Python will not run any part of the file.

**Common causes:** a missing colon at the end of an `if` or `for` line, unmatched parentheses, or a typo in a keyword.

````python
# This cell is supposed to error
# Missing colon after the if condition
x = 10
if x > 5
  print("x is greater than 5")
````

````python
# This cell is supposed to error
# what is the problem here?
ages = {
    'pam': 24,
    'jim': 24
    'michael': 43
}
print(f'Michael is {ages["michael"]} years old.')
````

The traceback will point directly at the line with the problem and sometimes even mark the exact position with a `^` character. Look at where Python says the error is -- it is usually right at or just after the spot where the syntax breaks down.

---

####**1.2b - Runtime Errors:**

These errors happen *while* the program is running and produce a traceback.

<br>

####**`IndentationError:`** wrong indentation

Python uses indentation to define code blocks. If the indentation is inconsistent or missing, Python raises an `IndentationError`.

- **Common causes:** mixing tabs and spaces, forgetting the indent of a function or loop, or accidentally adding extra indentation levels.

````python
# This cell is supposed to error
# The body of the function is not indented
def greet(name):
print(f"Hello, {name}")
````

---

####**`NameError:`** a variable or function does not exist

A `NameError` means Python cannot find a name -- usually because you misspelled a variable name, used a variable before assigning it, or forgot to define a function before calling it.

````python
# This cell is supposed to error
# city_name is misspelled as city_nmae
city_name = "Oakland"
print(city_nmae)
````

The error message tells you exactly which name Python could not find: `NameError: name 'city_nmae' is not defined`. Compare it to what you defined -- in this case, `city_name` vs `city_nmae`.

---

#### **`TypeError:`** the wrong type for an operation

A `TypeError` means you tried to do something with a value that does not support that operation -- for example, adding a string to a number, or calling something that is not a function.

- **Common causes:** forgetting to convert `input()` to `int` or `float`, passing the wrong type to a function, or confusing a variable with a function. Use `type()` to check your types to fix the issue.

````python
# This cell is supposed to error
# Cannot add a string and an integer directly
year = "2024"
years_from_now = 10
future_year = year + years_from_now
print(future_year)
````

The error message will say something like: `TypeError: can only concatenate str (not "int") to str`. Python is telling you exactly which types are incompatible.

- **Fix:** convert to the same type first.
```python
future_year = int(year) + years_from_now   # convert string to int
# or
future_year = year + str(years_from_now)   # convert int to string (gives '202410', probably not what you want)
```

---

####**`ValueError:`** the right type, but an impossible value

A `ValueError` means the type is correct but the value itself does not make sense for the operation -- for example, asking Python to convert the string `"hello"` to an integer.

````python
# This cell is supposed to error
# "hello" is a string (correct type for int()), but not a valid integer
user_input = "hello"
number = int(user_input)
print(number)
````

This comes up frequently when reading data from files or user input. A common pattern is to validate the input first:

```python
if user_input.isnumeric():
    number = int(user_input)
else:
    print(f"'{user_input}' is not a valid number")
```

---

####**`IndexError:`** accessing a position that does not exist in a list

An `IndexError` means you tried to access an element at a position (index) that is out of range for the list. Remember that Python lists are zero-indexed: a list with 5 elements has indices 0 through 4. Trying to access index 5 will raise this error.

````python
# This cell is supposed to error
cities = ["Oakland", "Berkeley", "San Francisco"]
print(cities[0])   # works: first element
print(cities[2])   # works: third (last) element
print(cities[3])   # error: there is no fourth element
````

The error message will say: `IndexError: list index out of range`.

---

####**`KeyError:`** accessing a dictionary key that does not exist

A `KeyError` is the dictionary equivalent of an `IndexError`. You tried to look up a key that is not in the dictionary.

````python
# This cell is supposed to error
city_populations = {
    "Oakland": 440646,
    "Berkeley": 124321,
}

print(city_populations["Oakland"])       # works
print(city_populations["San Francisco"]) # error: this key does not exist
````

The error message includes the key that was missing: `KeyError: 'San Francisco'`.

**Two safe alternatives to direct key access:**
```python
# Option 1: check before accessing
if "San Francisco" in city_populations:
    print(city_populations["San Francisco"])

# Option 2: use .get(), which returns None if the key is missing
pop = city_populations.get("San Francisco")   # returns None instead of erroring
pop = city_populations.get("San Francisco", 0) # returns 0 as a default instead
```

---


#### **`FileNotFoundError:`** the file does not exist at that path

A `FileNotFoundError` means Python could not find the file you asked it to open. This is extremely common in data work. Possible causes: the filename is misspelled, the file is in a different directory than you expect, or the file has not been created yet.

````python
# This cell is supposed to error
# The file 'data.csv' does not exist in the current directory
with open("data_that_does_not_exist.csv") as f:
    content = f.read()
````

When you see this error, the **first thing to do is confirm where Python is looking:**

```python
import os
print(os.getcwd())        # prints the current working directory
print(os.listdir())       # lists all files in the current directory
```

**Then check:** is your file actually there? Is the spelling exactly right (including uppercase/lowercase)?

##**Part 2: Sematic Errors - Bugs That Do Not Crash (Wrong Output)**

The second kind of bug is harder: **the program runs without any error, but produces the wrong result.** Python cannot help you here because as far as Python is concerned, nothing went wrong.

### 2.1 A Broken Program

This program asks the user a multiplication question, checking if their answer is correct. Run the cell and try answering with the right number.

````python
import random

x = random.randrange(1, 10)
y = random.randrange(1, 10)

answer = input(f"What's {x} multiplied by {y}? ")
expected = x * y

if answer == expected:
    print("Correct!")
else:
    print("That's incorrect")
````

Even when you type the right answer, the program always says "That's incorrect". No traceback, no error -- just wrong behavior.

---

### 2.2 Debugging with `print()`

The most common approach is to add `print()` statements to see what your variables actually hold while the program is running.

````python
import random

x = random.randrange(1, 10)
y = random.randrange(1, 10)

answer = input(f"What's {x} multiplied by {y}? ")
expected = x * y

# Debugging: look at what these variables actually hold
print("answer:", answer)
print("expected:", expected)

if answer == expected:
    print("Correct!")
else:
    print("That's incorrect")
````

Both values look the same -- for example, `5` and `5`. So why are they not equal? The problem is that `print()` shows you the **human-readable** version of a value, which hides the type.

---

### 2.3 The Real Problem: Types

Every value in Python has a **type**. Two of the most important for this bug:
- **`int`** -- a whole number, like `5` or `18`.
- **`str`** -- text, like `"5"` or `"18"`.

They look the same when printed, but Python treats them as different things. `"5"` (a string) is not equal to `5` (an integer):

````python
print("5" == 5)   # False: different types
print(5 == 5)     # True: both integers
````

The `input()` function **always returns a string**, no matter what the user types. Meanwhile, `x * y` produces an integer. So `answer` is a string and `expected` is an integer -- they will never be equal, even when the numbers match.

**Solution:** Use `type()` to check what type your variables are using to make sure you are comparing the right types.

````python
print(type("5"))   # Returns str
print(type(5))     # Returns int
````

‼️ **Pro-tip:** You can use an f-string feature to see both the variable name and its value at once. Adding = inside {} makes Python print the name and value automatically, which can help you spot type differences (like strings vs numbers).
- This is a **self-documenting expression** because the output documents itself -- it shows both the name and the value in one line

````python
answer = "18"    # string (as if it came from input())
expected = 18    # integer

# These two lines tell you everything you need to know
print(f"{answer = }")
print(f"{expected = }") # Added spaces around = for better formatting
````

Output:
```
answer = '18'
expected = 18
```

At a glance: `answer` is a string (note the quotes), `expected` is an integer. Bug found.

---

### 2.4 Fixing the Original Program

The fix is to convert the string from `input()` into an integer using `int()`:

````python
import random

x = random.randrange(1, 10)
y = random.randrange(1, 10)

# int() converts the string from input() into an integer
answer = int(input(f"What's {x} multiplied by {y}? "))
expected = x * y

# Confirm both are now integers before removing these debug lines
print(f"{answer = }")
print(f"{expected = }")

if answer == expected:
    print("Correct!")
else:
    print("That's incorrect")
````

> **Important habit: remove your debugging print statements** once you are done, before submitting or sharing your code.

> **Key fact to remember:** `input()` **always returns a string**. Any time you read a number from the user, convert it with `int()` or `float()` before doing math with it.

##**Part 3: Debugging Strategies**

### The scientific method

When you hit a bug, resist the urge to randomly change things. Instead, treat it like an experiment. Look at what the program actually did, form a hypothesis about *why*, make a prediction about what you will see if your hypothesis is right, and then test it. If the test confirms your hypothesis, fix it. If not, form a new hypothesis and repeat.

**Make one change at a time**. If you change three things at once and the bug disappears, you will not know which change fixed it -- or whether you accidentally introduced a new problem somewhere else.

Applied to the multiplication bug from Part 2:

- **Observe:** the program always says "That's incorrect" even when the answer looks right.
- **Hypothesize:** `input()` might return a string, so comparing it to an integer always fails.
- **Predict:** if I print both values with `f"{answer = }"`, the string will show quotes around it.
- **Experiment:** add the print, run the program, see `answer = '18'` vs `expected = 18`. Confirmed.
- **Fix:** wrap `input()` in `int()` and verify the output changes to `answer = 18`.

### Wolf fence debugging

A wolf is hiding somewhere in Alaska. Instead of searching everywhere, you build a fence down the middle and listen. If the wolf is on the right side, you hear it howl. If it's on the left side, the fence blocks the sound: silence.
Now you know which half it's in. Fence that half down the middle and listen again. Keep going until the area is so small the wolf has nowhere to hide.

Applied to code: drop a print statement in the middle of your program. If the output looks correct at that point, the bug is in the second half. Move the fence there. Keep narrowing until you find the exact line where things go wrong.

````python
import os

# Imagine this pipeline is broken somewhere -- you are not sure where.
# Plant a fence after each step and move it down until the output stops looking right.

files = os.listdir(".")
print("--- FENCE 1 ---", files[:3])     # looks reasonable? move the fence down

csv_files = [f for f in files if f.endswith(".csv")]
print("--- FENCE 2 ---", csv_files[:3]) # still good? move the fence down

# If FENCE 2 prints an empty list when you expected files, the bug is between FENCE 1 and FENCE 2.
# You no longer need to look at anything before FENCE 1 -- the wolf is in the second half.
````

### Rubber duck debugging

Get a rubber duck (or any object willing to listen). Explain your code to it, line by line, describing what each line is supposed to do. At some point you will say "and here it does..." and realize that what it actually does is not what you just said. That is the bug.

The reason this works is not magic: explaining code out loud forces you to pay attention to every line. Reading code silently lets your brain fill in what it *expects* to see. Speaking it aloud does not. If no duck is available, explain the code in writing, or to another student.

### Quick reference: practical checks

- **Check your working directory:** if files are not found, run `os.getcwd()` and `os.listdir()` to confirm Python is looking where you think.
- **Check variable types:** use `print(f"{var = }")` or `print(type(var))`. A value that looks right may have the wrong type.
- **Search the exact error message:** copy the last line of the traceback and search for it. Use the full message, not just the error type.
- **Check your assumptions:** when a bug seems impossible, print the variable you are most confident about. It is usually not what you think it is.

##**Summary**

**When your program crashes with an error:**
1. Go straight to the **bottom** of the traceback -- that is where the error type and message are.
2. Read the line above it -- that is the exact line of code that caused the crash.
3. Identify the error type using the reference table in Section 1.3.
4. If you do not understand the message, search for it verbatim online.

**When your program runs but gives wrong results:**
1. Add `print(f"{variable = }")` statements to see what your variables actually hold.
2. Check types with `type()` -- the value and the type together tell the whole story.
3. Remember that `input()` always returns a string. Convert with `int()` or `float()` before doing math.
4. Isolate the problem by testing one step at a time.

**The debugging print progression:**
```python
print("answer:", answer)           # basic -- shows value, hides type
print(f"{answer = }")              # best -- name, type, and value in one line
```
