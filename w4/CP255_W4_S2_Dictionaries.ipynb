## 1. What is a Dictionary?

A dictionary is a collection of **key-value pairs**. Think of it like a lookup table:

| Key | Value |
|-----|-------|
| "New York" | 8419600 |
| "Los Angeles" | 3980400 |
| "Chicago" | 2716000 |

- **Keys** must be unique and immutable (strings, numbers, tuples)
- **Values** can be anything (numbers, strings, lists, even other dictionaries)

## 2. Creating Dictionaries

### Method 1: Curly braces with key-value pairs

```python
# City populations - the most common way to create a dictionary
city_population = {
    "New York": 8419600,
    "Los Angeles": 3980400,
    "Chicago": 2716000,
    "Houston": 2328000,
    "Phoenix": 1690000
}

city_population
```

### Method 2: Using `dict()` constructor with tuples

```python
# Same data, different syntax
city_population_2 = dict([
    ("New York", 8419600),
    ("Los Angeles", 3980400),
    ("Chicago", 2716000),
    ("Houston", 2328000),
    ("Phoenix", 1690000)
])

city_population_2
```

### Method 3: Using `zip()` to combine two lists

This is handy when you have separate lists of keys and values.

```python
cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
populations = [8419600, 3980400, 2716000, 2328000, 1690000]

city_population_3 = dict(zip(cities, populations))
city_population_3
```

## 3. Accessing Values

### Using bracket notation `[]`

```python
# Get the population of Chicago
city_population['Chicago']
```

```python
# Use in an f-string
print(f"Population of Chicago: {city_population['Chicago']:,}")
```

```python
city_population['Oakland']
```

### Watch out: KeyError

If you try to access a key that doesn't exist, you get a `KeyError`:

```python
# This will raise a KeyError - uncomment to see
# city_population['San Francisco']
```

### Using `.get()` for safe access

The `.get()` method returns `None` (or a default value) instead of raising an error:

```python
# Returns None if key doesn't exist
result = city_population.get('San Francisco')
print(result)
```

```python
# Provide a default value
result = city_population.get('San Francisco', 0)
print(result)
```

```python
# Practical example: counting occurrences
inventory = {"apple": 100, "orange": 80, "banana": 100}

# How many mangoes? (doesn't exist)
mango_count = inventory.get("mango", "i dont have it")
print(f"Mangoes in stock: {mango_count}")
```

## 5. Adding and Modifying Entries

### Add a new key-value pair

```python
# Add San Diego
city_population['San Diego'] = 1424000
city_population
```

### Update an existing value

```python
# Update New York's population
city_population['New York'] = 8500000
city_population
```

### Using `.update()` to merge dictionaries

```python
# Add multiple cities at once
more_cities = {
    "Dallas": 1341000,
    "San Jose": 1035000
}

city_population.update(more_cities)
city_population
```

---

## 6. Removing Entries

### Using `.pop()` - removes and returns the value

```python
# Remove Dallas and get its value
dallas_pop = city_population.pop('Dallas')
print(f"Removed Dallas with population {dallas_pop:,}")
print(city_population)
```

### Safe removal with default

```python
# Won't raise error if key doesn't exist
result = city_population.pop('Miami', 0)
print(f"Miami population (or 0 if not found): {result}")
```

### Using `del` - just removes, doesn't return

```python
del city_population['San Jose']
city_population
```

---

## 7. Looping Over Dictionaries

### Loop over keys (default behavior)

```python
for city in city_population:
    print(city)
```

### Loop over key-value pairs with `.items()`

```python
for city, population in city_population.items():
    print(f"{city}: {population:,}")
```

### Building a dictionary in a loop

```python
# Create a dictionary of squares
squares = {}

for num in range(1, 6):
    squares[num] = num ** 2

squares
```

## 8. Checking Membership

Use `in` to check if a key exists:

```python
print('Chicago' in city_population)      # True
print('Miami' in city_population)        # False
```

```python
# Practical use in a conditional
city = 'Boston'

if city in city_population:
    print(f"{city} population: {city_population[city]:,}")
else:
    print(f"{city} not found in our data")
```

## 9. Connection to Pandas

Dictionaries are how we often create DataFrames:

```python
import pandas as pd

# Dictionary where keys become column names
data = {
    'city': ['Oakland', 'Berkeley', 'San Francisco'],
    'population': [433031, 124321, 874961],
    'area_sqmi': [78.0, 17.7, 46.9]
}

df = pd.DataFrame(data)
df
```

```python
# Convert DataFrame back to dictionary
df.to_dict()
```

## 10. Nested Dictionaries & GeoJSON Preview

Dictionaries can contain other dictionaries. This is how GeoJSON works:

```python
# A simplified GeoJSON-like structure
feature = {
    "type": "Feature",
    "properties": {
        "name": "Downtown Oakland",
        "population": 25000,
        "zoning": "C-45"
    },
    "geometry": {
        "type": "Point",
        "coordinates": [-122.2711, 37.8044]
    }
}

# Access nested values
print(f"Name: {feature['properties']['name']}")
print(f"Coordinates: {feature['geometry']['coordinates']}")
```

## Summary

| Operation | Syntax |
|-----------|--------|
| Create | `d = {'key': value}` or `dict()` |
| Access | `d['key']` or `d.get('key', default)` |
| Add/Update | `d['key'] = value` |
| Remove | `d.pop('key')` or `del d['key']` |
| Get keys | `d.keys()` |
| Get values | `d.values()` |
| Get pairs | `d.items()` |
| Check membership | `'key' in d` |
| Merge | `d.update(other_dict)` |

---

## Key Takeaways

1. **Use `.get()` for safe access** - avoids KeyError
2. **Use `.items()` for looping** - gives you both key and value
3. **Dictionaries → DataFrames** - keys become column names
4. **GeoJSON is nested dictionaries** - understanding dicts helps with spatial data
