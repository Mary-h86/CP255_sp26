## Part 2: Variables - Giving Names to Things

### What is a Variable?

A variable is a name that refers to a value. It lets you reuse data without rewriting it. The variable tells you how to find the thing you want, but it is not the thing itself.

Think of it as a label you can assign to an object. You can assign multiple labels to an object, all of them are referencing the same thing.

> **In Python, variables and data structures don't contain objects. They are pointers (references) to where the object is stored on the memory (its a mouthful but lets just leave it at that for now, it will become more clear as we go on). This fact is both commonly overlooked and tricky to internalize.**

You can happily use Python for years without really understanding this concept, but this knowledge can certainly help alleviate many common Python gotchas. So now that we have just started, lets learn this correctly from the beginning.

### Why Use Variables?
Variables allow programmers to:

- Reference and reuse values: Avoid the need to retype values throughout the code.
- Make code readable: Meaningful labels like name or age clarify the purpose of the data being used.
- Easily modify references: Change which object a variable points to without affecting other parts of the program.

### The Assignment Operator: `=`

The single equals sign `=` is called the **assignment operator**. It does NOT mean "equals" in the mathematical sense. Instead, it means:

> "Take the value on the right and reference it to the variable on the left."

```python
variable_name = value
```

### Variable Naming Rules
:
A variable name should show the intended use!

Python has rules about what you can name your variables:

1. Must start with a letter or underscore (`_`)
2. Can contain letters, numbers, and underscores
3. Cannot contain spaces or special characters
4. Cannot be a Python keyword (like `if`, `for`, `while`, etc.)
5. Are case-sensitive (`Age` and `age` are different)


**Best Practice:** Use descriptive names in `snake_case`

| ✅ Valid | ❌ Invalid | Why Invalid |
|----------|-----------|-------------|
| `station_name` | `station-name` | Hyphens not allowed |
| `station2` | `2station` | Can't start with number |
| `_private` | `class` | `class` is a reserved word |
| `totalRidership` | `total ridership` | No spaces allowed |

### Convention: Use `snake_case`

In Python, we use **snake_case** for variable names: all lowercase, with underscores separating words.

```python
# Good (snake_case)
daily_ridership = 15000
average_trip_length = 12.5

# Works but not conventional
DailyRidership = 15000  # This is "PascalCase" - used for classes
dailyRidership = 15000  # This is "camelCase" - used in other languages
```

## Part 4: Operators

Now that we have variables, we want to use them.

**Operators** let us perform calculations, comparisons, and more.

### Arithmetic Operators

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division | `5 / 3` | `1.666...` |
| `//` | Floor Division | `5 // 3` | `1` |
| `%` | Modulo (remainder) | `5 % 3` | `2` |
| `**` | Exponent | `5 ** 3` | `125` |

````python
# Let's calculate total daily ridership at Downtown Berkeley

daily_entries = 4521
daily_exits = 4892

# Total people using the station
total_daily_traffic = daily_entries + daily_exits
print("Total daily traffic:", total_daily_traffic)

# Average (people coming vs going)
average_flow = total_daily_traffic / 2
print("Average flow:", average_flow)
````

````python
# More calculations: annual ridership estimate

weekday_riders = 4521
weekend_riders = 2100  # Lower on weekends

# 52 weeks per year, 5 weekdays and 2 weekend days per week
annual_estimate = (weekday_riders * 5 + weekend_riders * 2) * 52

print("Estimated annual entries:", annual_estimate)
````

### Floor Division (`//`) and Modulo (`%`)

These two operators are super useful for urban planning calculations:

- **Floor division (`//`)**: Divides and rounds DOWN to the nearest integer
- **Modulo (`%`)**: Gives you the REMAINDER after division

````python
# Example: How many full BART trains can we fill?

waiting_passengers = 1547
train_capacity = 200  # passengers per train

# How many FULL trains?
full_trains = waiting_passengers // train_capacity
print("Full trains needed:", full_trains)

# How many passengers left over?
remaining_passengers = waiting_passengers % train_capacity
print("Passengers waiting for next train:", remaining_passengers)
````

### String Operators: `+` and `*`

The `+` and `*` operators work differently with strings:

- `+` **concatenates** (joins) strings together
- `*` **repeats** a string

````python
# String concatenation

first_name = "Downtown"
last_name = "Berkeley"

# Join them together with a space
full_name = first_name + " " + last_name
print(full_name)

# String repetition
divider = "-" * 30
print(divider)
````

### Comparison Operators

Comparison operators compare two values and return a **boolean** (`True` or `False`).

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal | `5 >= 5` | `True` |
| `<=` | Less than or equal | `5 <= 3` | `False` |

#### **IMP** Common Mistake: `=` vs `==`

- `=` is for **assignment** (Creating a variable by assigning a value to it)
- `==` is for **comparison** (checking if two values are equal)

````python
# Comparison examples

berkeley_ridership = 4521
oakland_ridership = 8234

# Is Berkeley busier than Oakland?
print("Is Berkeley busier?", berkeley_ridership > oakland_ridership)

# Is Oakland busier?
print("Is Oakland busier?", oakland_ridership > berkeley_ridership)

# Are they equal?
print("Same ridership?", berkeley_ridership == oakland_ridership)
````

## Part 5: Accessing Objects Built-in Menu: The Dot (`.`) - Accessing Methods

### What is a Method?

Each object comes with its own built-in menu. Like right-clicking an item on your desktop to see available actions, a method is one item on an object’s built-in menu; an action that only works for that kind of object.
More formally, a **method** is a function that "belongs to" a specific type of data. It's an action that the can be performed on that specific type of objects.

Think of it this way:
- A **variable** is a noun (a thing - an object)
- A **method** is a verb (an action the object can do)

### The Dot (`.`) Operator

The **dot** connects an object to its methods. When you see:

```python
station_name.upper()
```

Read this as: "Hey `station_name`, do your `upper` action!"

The dot is like saying "**'s**" in English:
- `station_name.upper()` → "station_name**'s** upper method"

### Why the Parentheses?

Methods always have parentheses `()` after them because they're **actions**, you're telling Python to **DO something**.

The parentheses are where you can pass additional information (called **arguments**) if needed.

````python
# String methods

station = "downtown berkeley"

# .upper() - convert to uppercase
print(station.upper())

# .lower() - convert to lowercase
print(station.lower())

# .title() - capitalize first letter of each word
print(station.title())

# .capitalize() - capitalize only the first letter
print(station.capitalize())
````

### Methods with Arguments

Some methods need additional information to do what you’re asking.
When a method is defined, it lists parameters (the placeholders) that it needs.
When you call the method, you provide arguments (the actual values) inside the parentheses, either by position or by keyword (more later!).

````python
# The .replace() method needs TWO arguments:
# 1. What to find
# 2. What to replace it with

station = "12th St. Oakland City Center"

# Replace "St." with "Street"
full_name = station.replace("St.", "Street")
print(full_name)
````

````python
# The .count() method counts occurrences

bart_lines = "Blue, Orange, Red, Yellow, Green, Blue, Orange"

# How many times does "Blue" appear?
blue_count = bart_lines.count("Blue")
print("Blue appears", blue_count, "times")
````

### Different Types Have Different Methods

Each data type has its own set of methods that make sense for that type:

- Strings have methods like `.upper()`, `.lower()`, `.replace()`
- Numbers have different methods (we'll see more later)
- Lists and dictionaries have their own methods (coming in Week 3!)

**The dot connects the data to methods that work with that type of data.**

````python
# Strings have .upper()
text = "hello"
print(text.upper())  # Works!

# Numbers don't have .upper()
number = 42
# print(number.upper())  # This would cause an ERROR!

# But numbers have other useful methods
# For example, floats can check if they're a whole number:
price = 10.0
print("Is price a whole number?", price.is_integer())
````

## Part 6: Type Conversion - Changing Types

Sometimes you need to convert data from one type to another. Python provides functions for this:

| Function | Converts to | Example |
|----------|------------|--------|
| `str()` | String | `str(42)` → `"42"` |
| `int()` | Integer | `int("42")` → `42` |
| `float()` | Float | `float("3.14")` → `3.14` |
| `bool()` | Boolean | `bool(1)` → `True` |

````python
# Converting for string concatenation

station = "Downtown Berkeley"
ridership = 4521

# This won't work:
# message = station + " has " + ridership + " daily entries"

# We need to convert the number to a string:
message = station + " has " + str(ridership) + " daily entries"
print(message)
````

````python
# Converting string input to numbers

# Imagine this came from a data file as text
ridership_text = "4521"

# We can't do math with a string!
# doubled = ridership_text * 2  # This repeats the string: "45214521"

# Convert to int first:
ridership_num = int(ridership_text)
doubled = ridership_num * 2
print("Doubled ridership:", doubled)
````
